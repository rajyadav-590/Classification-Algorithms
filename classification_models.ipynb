```python
X = df.drop(columns=['HeartDisease'], axis= 1)
X
```

```python
y = df['HeartDisease']
y
```

```python
# one hot encoding
X_one_encoder = pd.get_dummies(X)
X_one_encoder = X_one_encoder.astype(int)
X_one_encoder
```

```python
# scaling one hot encoded data
from sklearn.preprocessing import StandardScaler
scaler =StandardScaler()
X_scaled = scaler.fit_transform(X_one_encoder)
X_scaled
```

```python
# spliting one hot encoding scaled data
from sklearn.model_selection import train_test_split
X_train_one_encoder_scaled, X_test_one_encoder_scaled, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state= 42)
X_train_one_encoder_scaled
```

```python
X_test_one_encoder_scaled
```

```python
# label encoding
from sklearn.preprocessing import LabelEncoder
X_label = X.apply(LabelEncoder().fit_transform)
X_label
```

```python
# spliting without scaling
from sklearn.model_selection import train_test_split
X_train_label, X_test_label, y_train, y_test = train_test_split(X_label, y, test_size=0.2, random_state=42)
X_train_label
```

```python
# scaling X_label then spliting
X_label_scaled = scaler.fit_transform(X_label)
X_label_scaled
```

```python
# spliting labeled and scaled data
X_train_label_scaled, X_test_label_scaled, y_train, y_test = train_test_split(X_label_scaled, y, test_size=0.2, random_state=42)
X_train_label_scaled
```

# training classification models 

## 1. logistic regression

```python
# using one hot encoding
from sklearn.linear_model import LogisticRegression
model_one_hot = LogisticRegression()
model_one_hot.fit(X_train_one_encoder_scaled, y_train)
y_pred_one_hot = model_one_hot.predict(X_test_one_encoder_scaled)
y_pred_one_hot
```

```python
# evaluating predictions
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
accuracy_score(y_test, y_pred_one_hot)
```

```python
confusion_matrix(y_test, y_pred_one_hot)
```

```python
print(classification_report(y_test, y_pred_one_hot))
```

```python
# using label encoding without scaling 
model_label = LogisticRegression()
model_label.fit(X_train_label, y_train)
y_pred_xlabel = model_label.predict(X_test_label)
y_pred_xlabel
```

```python
accuracy_score(y_test,y_pred_xlabel)
```

```python
confusion_matrix(y_test, y_pred_xlabel)
```

```python
print(classification_report(y_test, y_pred_xlabel))
```

```python
# using label encoding with scaling 
model_label_scaled = LogisticRegression()
model_label_scaled.fit(X_train_label_scaled, y_train)
y_pred_xlabel_scaled = model_label_scaled.predict(X_test_label_scaled)
y_pred_xlabel_scaled
```

```python
accuracy_score(y_test, y_pred_xlabel_scaled)
```

```python
confusion_matrix(y_test, y_pred_xlabel_scaled)
```

```python
print(classification_report(y_test, y_pred_xlabel_scaled))
```
